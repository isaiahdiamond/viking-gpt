```python
%load_ext autoreload
%autoreload 2
from engine import Value
from nn import MLP
import json
import random

# Loading normalized banknote data 
with open("data/banknote_split.json") as f:
    d = json.load(f)

X_train, Y_train, X_test, Y_test = d["X_train"], d["Y_train"], d["X_test"], d["Y_test"]
```

```python
# Creating a 2-layer MLP that takes 4 inputs from X_trian and produces 1 ypred 

mlp = MLP(4,[8,8,1])
```

```python
# -------Training Loop-----------

batch_size = 32
lr = .03 
nsteps = 500

for k in range(nsteps):
    # Batch selection 
    idx = random.sample(range(len(X_train)), batch_size)   # 32 distinct rows
    xb = [X_train[i] for i in idx]
    yb = [Y_train[i] for i in idx]

    # Forward pass
    ypred = [mlp(row) for row in xb]
    loss = sum([(yout - ygt) ** 2 for ygt, yout in zip(yb, ypred)])

    # Backward pass
    for p in mlp.parameters():
        p.grad = 0.0

    loss.backward()

   # Update
    for p in mlp.parameters():
        p.data += -lr * p.grad

    if k % 25 == 0:
        print(f"Loop: {k} , Loss: {loss.data}")

```

```python
# Running the nn on the test data to get it's predictions on unseen data.

test_pred = [mlp(row) for row in X_test]

# Modify its predictions to a binary +/- 1
final_pred = [1.0 if v.data > 0 else -1.0 for v in test_pred]

final_pred
```
