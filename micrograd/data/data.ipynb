```python
import urllib.request, random

# importing banknote data
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"
rows = [list(map(float, line.split(','))) for line in urllib.request.urlopen(url).read().decode().split()]
random.seed(42)
random.shuffle(rows)


# Each banknote has these features [ variance, skewness, kurtosis, entropy, class ]
X = [r[:4] for r in rows]
Y = [1.0 if r[4] == 1 else -1.0 for r in rows]   # -1/+1 to match tanh's output range

# Spliting data into train/test

X_train = X[:1098]
Y_train = Y[:1098]

X_test = X[1098:]
Y_test = Y[1098:]
```

```python
# Normalizing the data set
import math 

col_1 = 0
col_2 = 0 
col_3 = 0
col_4 = 0

for row in X_train:
    col_1 += row[0]
    col_2 += row[1]
    col_3 += row[2]
    col_4 += row[3]

means = [col_1/len(X_train),col_2/len(X_train),col_3/len(X_train),col_4/len(X_train)]

std_1 = 0
std_2 = 0
std_3 = 0
std_4 = 0

for row in X_train:
    std_1 += (row[0] - means[0]) ** 2  
    std_2 += (row[1] - means[1]) ** 2  
    std_3 += (row[2] - means[2]) ** 2  
    std_4 += (row[3] - means[3]) ** 2  

stds = [
    math.sqrt(std_1/len(X_train)),
    math.sqrt(std_2/len(X_train)),
    math.sqrt(std_3/len(X_train)),
    math.sqrt(std_4/len(X_train))
]
```

```python
print(means)
print(stds)
```

```python
# Normalizing the whole set 
X_test_norm = []
X_train_norm = []

for row in X_train:
    X_train_norm.append([((row[0] - means[0]) / stds[0]),((row[1] - means[1]) / stds[1]),((row[2] - means[2]) / stds[2]),((row[3] - means[3]) / stds[3])])

for row in X_test:
    X_test_norm.append([((row[0] - means[0]) / stds[0]),((row[1] - means[1]) / stds[1]),((row[2] - means[2]) / stds[2]),((row[3] - means[3]) / stds[3])])
```

```python
# saving normalized values
import json

with open("banknote_split.json", "w") as f:
    json.dump({
        "X_train": X_train_norm,
        "Y_train": Y_train,
        "X_test":  X_test_norm,
        "Y_test":  Y_test,
        "means":   means,   # to normalize any new note the same way later
        "stds":    stds,
    }, f)
```
